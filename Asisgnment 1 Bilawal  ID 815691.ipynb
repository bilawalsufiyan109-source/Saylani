Take marks and print "Pass" if marks ≥ 40, else "Fail"

```python
marks = int(input("Enter Marks:"))
if marks >= 40:
    print("Exam pass")
else:
    print("Exam Fail")
```

Check if a number is between 10 and 50 using and.

```python
num = int(input("Enter a number: "))

if num >= 10 and num <= 50:
    print("Number is between 10 and 50")
else:
    print("Number is not between 10 and 50")
```

Check if a number is divisible by 3 OR 5 using or 

```python
num1 = int(input("Enter divisble NUM:"))

if num1 % 3 == 0 or num1 % 5 == 0:
    print("its divided by 3 or 5")
else:
    print("its not divislble")
```
