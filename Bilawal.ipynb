## 7) Compound Interest:

```python
P = 1000
r = 0.05
n = 1
t = 2

A = P * (1 + r / n) ** (n * t)

print("Total Amount =", A)
```

## 11) Quadratic Equation Roots:

```python
import math

a = 1
b = -5
c = 6

D = b ** 2 - 4 * a * c

x1 = (-b + math.sqrt(D)) / (2 * a)
x2 = (-b - math.sqrt(D)) / (2 * a)

print("First root =", x1)
print("Second root =", x2)
```
